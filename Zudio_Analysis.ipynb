### Create Derived Metrics

```python
df['Total_Sales'] = df['Price'] * df['Quantity']
df['Profit_Margin'] = (df['Sales_Profit'] / df['Total_Sales']) * 100
print(df[['Total_Sales', 'Profit_Margin']])
```

### Trend Analysis

```python
df['Store_Open_Date'] = pd.to_datetime(df['Store_Open_Date'], format='%d-%m-%Y')
df['Year'] = df['Store_Open_Date'].dt.year
Sales_Trend = df.groupby(['Year'])['Total_Sales'].sum().reset_index()
print(Sales_Trend)
```

```python
plt.figure()
plt.plot(Sales_Trend['Year'], Sales_Trend['Total_Sales'])
plt.title("Yearly Sales Trend")
plt.xlabel("Year")
plt.xticks(Sales_Trend['Year'])
plt.ylabel('Total Sales')
plt.show()
```

### YoY / MoM Growth

```python
yearly_sales = df.groupby("Year")["Total_Sales"].sum().reset_index()

yearly_sales["prev_year_sales"] = yearly_sales["Total_Sales"].shift(1)

yearly_sales["growth_rate"] = (
    (yearly_sales["Total_Sales"] - yearly_sales["prev_year_sales"]) 
    / yearly_sales["prev_year_sales"]
) * 100
print(yearly_sales["growth_rate"])
```

```python
plt.figure()
plt.plot(yearly_sales["Year"], yearly_sales["growth_rate"])
plt.title("Year-over-Year Growth (%)")
plt.xlabel("Year")
plt.xticks(yearly_sales['Year'])
plt.ylabel("Growth %")
plt.show()
```

### Product-Level Analysis

```python
product_analysis = df.groupby("Clothing_Type").agg({
    "Total_Sales": "sum",
    "Sales_Profit": "sum"
}).reset_index().sort_values(by="Total_Sales", ascending=False)
print(product_analysis)
```

```python
top_products = product_analysis.head(10)

plt.figure()
plt.bar(top_products["Clothing_Type"], top_products["Total_Sales"])
plt.title("Top 10 Products by Sales")
plt.xlabel("Product")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.show()
```
